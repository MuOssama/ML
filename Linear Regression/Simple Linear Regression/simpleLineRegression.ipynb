```python
from sklearn.linear_model import LinearRegression
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Training data
X = np.array([
    [50, 2],
    [70, 2],
    [80, 3],
    [100, 3],
    [120, 4]
])

y = np.array([
    50000,
    70000,
    85000,
    110000,
    135000
])
```

```python
model = LinearRegression()
model.fit(X,y)
```

```python
prediction_ip = np.array([[110,3],[130,3]])
prediction_op = model.predict(prediction_ip)
model.fit(prediction_ip,prediction_op)
```

```python
plt.scatter(prediction_ip[:,0],prediction_op,label="predictions")
plt.scatter(X[:,0],y,label="true")
#X[:,0] means get all of 0th coulmn

plt.xlabel("Area")
plt.ylabel("Price")
plt.title("Area vs House Price")
plt.show()
```
